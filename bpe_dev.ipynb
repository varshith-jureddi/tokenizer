# Iterative Version

```python
# using longer text
text = """హలో నమస్కారం arogyanga undataniki moodu chitkalu ivvandi.1. samathulya mariyu poshakamaina aaharam thinandi: mi bhojanamlo vividha rakala pandlu mariyu kuragayalu, sannani protean, trindhanyalu mariyu arogyakaramaina kovvulu unnaayani nirdharimchukondi. idi mee syareeraaniki uttamanga panicheyadaniki avasaramaina poshakalanu andinchamlo sahayapaduthundi mariyu dirghakalika vyadhulanu nivarinchada sahayapaduthundi."""
tokens=text.encode("utf-8")
tokens = list(map(int,tokens))
```

```python
def get_stats(ids):
    counts={}
    for pair in zip(ids,ids[1:]):
        counts[pair]=counts.get(pair,0)+1
    return counts

def merge(ids,idx,pair): #ids-tokens list,idx-new token or index, pair - top pair
    newids=[]
    i=0
    added=0
    while i<len(ids):
        if i<len(ids)-1 and ids[i]==pair[0] and ids[i+1]==pair[1]:
            newids.append(idx)
            i+=2
        else:
            newids.append(ids[i])
            i+=1
    return newids

vocab_size = 276 # vocabulary size wanted
total_merges = vocab_size-256 # initial vocab is 256(256 bytes)
ids=list(tokens)

merges={} #to keep track of merges - (int,int)->int
# merging - adding new tokens for bpe
for i in range(total_merges):
    stats=get_stats(ids)
    pair=max(stats,key=stats.get)
    idx=256+i
    print(f"merging {pair} into new token - {idx}")
    merges[pair]=idx
    ids=merge(ids,idx,pair)
    
```

```python
print("original",len(tokens))
print("compressed",len(ids))
print("compression ratio",f"{len(tokens)/len(ids):.2f}")
      
```

# Tokenizer

```python
# vocabulary
vocab = {idx : bytes([idx]) for idx in range(256)}

#adding merges
for (p,q),i in merges.items():
    vocab[i]=vocab[p]+vocab[q]
```

    bytes(iterable_of_ints) -> bytes

    bytes() -> empty bytes object

## Decoder

```python
# DEcode
def decode(ids):
    tokens=b"".join(vocab[i] for i in ids)
    text = tokens.decode("utf-8",errors="replace")
    # print(text)
    return text
# print(decode([97]))
```

## Encoder

```python
# def get_pair(p):
#     return merges.get(p,float("inf")
                      
def encode(text):
    tokens=list(text.encode("utf-8"))
    # while True:
    while len(tokens)>1:
        stats=get_stats(tokens)
        pair=min(stats,key= lambda p: merges.get(p,float("inf")))
        if pair not in merges:
            break
        idx=merges[pair]
        tokens=merge(tokens,idx,pair)
    return tokens
print(encode("namaskaram"))
```
